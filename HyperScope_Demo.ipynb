## Curve view

```python
x = np.linspace(-5, 5, 301)
curve = np.exp(-(x-1)**2 / 1.5) + 0.05*np.sin(8*x)
curve_view = open_view(curve, axes=[x], title='Demo curve', units=['cm⁻¹', 'a.u.'])
curve_view.show()
```

```python
curve_result = curve_view.extract()
curve_view.save_numpy('curve_result.npz')
curve_result['profiles'].keys(), curve_view.metadata()
```

## 2D Image view

```python
y, x2 = np.mgrid[-3:3:80j, -4:4:100j]
image = np.exp(-((x2-1)**2 + (y+0.5)**2)) + 0.6*np.exp(-((x2+1.5)**2 + (y-1)**2)/0.4)
image_view = open_view(image, axes=[x2[0], y[:,0]], title='Demo image')
image_view.show()
```

```python
x_profile = image_view.profiles()['x']
display_image = image_view.images()['display']
image_view.save_image('image_display.png')
x_profile.shape, display_image.shape
```

## Hyperspectral view

```python
yy, xx, zz = np.mgrid[-2:2:35j, -3:3:45j, 400:800:90j]
cube = np.exp(-((xx-0.8)**2 + (yy+0.4)**2)) * np.exp(-((zz-650)/35)**2)
cube += 0.7*np.exp(-((xx+1.2)**2 + (yy-0.8)**2)/0.5) * np.exp(-((zz-520)/25)**2)
hyper_view = open_view(cube, axes=[xx[0,:,0], yy[:,0,0], zz[0,0,:]], title='Demo hyperspectral image')
hyper_view.show()
```

```python
profiles = hyper_view.profiles()
images = hyper_view.images()
scipy_ready = hyper_view.to_scipy()
hyper_view.save_numpy('hyperspectral_result.npz')
hyper_view.save_scipy('hyperspectral_result.mat')
hyper_view.save_image('hyperspectral_display.png')
profiles.keys(), images.keys(), scipy_ready.shape
```
